# Task 3. Finite differences

Write a function that calculates finite differences:

$$\frac{df}{dx}(x_i) \approx \frac{f(x_{i + 1}) - f(x_{i})}{x_{i + 1} - x_i}$$

The inputs to this function are:
- values of the function $f(x_i)$ as `numpy.array` of shape $N$
- arguments $x_i$ as `numpy.array` of shape $N$

The output should be a `numpy.array` with approximate derivatives of shape $N - 1$.

````python
# Edit this cell in task.ipynb.
def finite_differences(f, x):
    df_x = f[1:] * 0  # fallback: correct shape (N-1), not the real values
    ## YOUR CODE HERE

    df_x = np.diff(f) / np.diff(x)

    return df_x
````

# 2D Derivatives and Gradients

Below are 2D functions. Find their derivatives with respect to $x$ and $y$, and return them as a pair of NumPy maps.

Note that in 2D, the `x` and `y` coordinates are 2D maps. Every number in these maps is the value of the $x$ or $y$ coordinate at that point.

`f` is also a map of values of the function $f$ for the corresponding coordinates $x$ and $y$.

Example:

```
x = np.array([
    [1, 2],
    [1, 2]
])

y = np.array([
    [1, 1],
    [2, 2]
])

f = x ** 2 + y ** 2
``` 

````python
def f1(x, y):
    res = np.exp(-0.1 * (x**2 + y**2))
    return res

def f2(x, y):
    res = np.sin(x) * np.cos(y)
    return res

def f3(x, y):
    res = np.sin(x) + np.cos(y)
    return res

def f4(x, y):
    res = -np.exp(-0.2 * (x**2 + y**2)) + 0.5 * np.sin(3*x) * np.sin(3*y)
    return res

def f5(x, y):
    res = np.sin(x * 2) * np.cos(y * 2) + 0.2 * np.sin(5*x) * np.cos(5*y)
    return res

def f6(x, y):
    res = np.sin(5*x) * np.sin(5*y)
    return res
````

# Task 4. Find gradients of 2D functions

Write functions that calculate the gradients of the functions above. Each function should return a tuple of maps: $\frac{\partial f}{\partial x}$, $\frac{\partial f}{\partial y}$

````python
# Edit this cell in task.ipynb.
def grad_f1(x, y):
    res = (x, y)
    ## YOUR CODE HERE

    f = np.exp(-0.1 * (x ** 2 + y ** 2))
    res = (-0.2 * x * f, -0.2 * y * f)

    return res

def grad_f2(x, y):
    res = (x, y)
    ## YOUR CODE HERE

    res = (np.cos(x) * np.cos(y), -np.sin(x) * np.sin(y))

    return res

def grad_f3(x, y):
    res = (x, y)
    ## YOUR CODE HERE

    res = (np.cos(x), -np.sin(y))

    return res

def grad_f4(x, y):
    res = (x, y)
    ## YOUR CODE HERE

    e = np.exp(-0.2 * (x ** 2 + y ** 2))
    dx = 0.4 * x * e + 1.5 * np.cos(3 * x) * np.sin(3 * y)
    dy = 0.4 * y * e + 1.5 * np.sin(3 * x) * np.cos(3 * y)
    res = (dx, dy)

    return res

def grad_f5(x, y):
    res = (x, y)
    ## YOUR CODE HERE

    dx = 2 * np.cos(2 * x) * np.cos(2 * y) + np.cos(5 * x) * np.cos(5 * y)
    dy = -2 * np.sin(2 * x) * np.sin(2 * y) - np.sin(5 * x) * np.sin(5 * y)
    res = (dx, dy)

    return res

def grad_f6(x, y):
    res = (x, y)
    ## YOUR CODE HERE

    res = (5 * np.cos(5 * x) * np.sin(5 * y), 5 * np.sin(5 * x) * np.cos(5 * y))

    return res
````

# Task 5. 2D finite differences

For the map $f$ and for two maps of $x$ and $y$, write a function that calculates an approximate derivative by $x$ and $y$ using finite differences:

$\frac{\partial f}{\partial x} \approx \frac{f_{k,l + 1} - f_{k, l}}{x_{k, l + 1} - x_{k, l}}$

$\frac{\partial f}{\partial y} \approx \frac{f_{k + 1,l} - f_{k, l}}{y_{k+1, l} - y_{k, l}}$

Here $k$ is the row index and $l$ is the column index of the `np.meshgrid` maps: $x$ changes with the column index $l$, and $y$ changes with the row index $k$.

````python
# Edit this cell in task.ipynb.
def finite_differences_2d(f, x, y):
    df_x = f[:, 1:] * 0  # fallback: correct shapes, not the real values
    df_y = f[1:, :] * 0
    ## YOUR CODE HERE

    df_x = np.diff(f, axis=1) / np.diff(x, axis=1)
    df_y = np.diff(f, axis=0) / np.diff(y, axis=0)

    return df_x, df_y
````
